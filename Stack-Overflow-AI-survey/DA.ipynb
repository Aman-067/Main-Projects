### Making final changes --> 
* Keeping some original numerical columns that have outliers and some unreal values.
* Making new columns with applied conditions for operations that can get affected by those unreal outlier values.
* changing the dtypes of columns to manage the size of data.

```python
data['Salary_USD'] = data['Salary_USD'].astype('float32')
data['Job_satisfaction'] = data['Job_satisfaction'].astype('Int8')
data['Work_exp'] = data['Work_exp'].astype('Int8')
data['Years_coding'] = data['Years_coding'].astype('Int8')
```
